```python
import pandas as pd
```

```python
path = 'runs/pose/train/results.csv'
df = pd.read_csv(path)
df.head()
```

```python
df.columns
```

```python
import matplotlib.pyplot as plt

def plot_training_losses(train_box_loss, train_pose_loss, train_class_loss, 
                        val_box_loss, val_pose_loss, val_class_loss):
    """
    Plota box loss, pose loss e class loss de treino e validação em uma única figura
    """
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    # Box Loss
    axes[0].plot(train_box_loss, label='Train Box Loss', color='blue')
    axes[0].plot(val_box_loss, label='Val Box Loss', color='red')
    axes[0].set_title('Box Loss')
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Loss')
    axes[0].legend()
    axes[0].grid(True)
    
    # Pose Loss
    axes[1].plot(train_pose_loss, label='Train Pose Loss', color='blue')
    axes[1].plot(val_pose_loss, label='Val Pose Loss', color='red')
    axes[1].set_title('Pose Loss')
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Loss')
    axes[1].legend()
    axes[1].grid(True)
    
    # Class Loss
    axes[2].plot(train_class_loss, label='Train Class Loss', color='blue')
    axes[2].plot(val_class_loss, label='Val Class Loss', color='red')
    axes[2].set_title('Class Loss')
    axes[2].set_xlabel('Epoch')
    axes[2].set_ylabel('Loss')
    axes[2].legend()
    axes[2].grid(True)
    
    plt.tight_layout()
    plt.show()
```

```python
plot_training_losses(df['train/box_loss'], df['train/pose_loss'], df['train/cls_loss'],
                        df['val/box_loss'], df['val/pose_loss'], df['val/cls_loss'])
```
